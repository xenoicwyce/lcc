```python
import json
import numpy as np
import networkx as nx

from pathlib import Path

from qiskit_optimization.converters import QuadraticProgramToQubo

from lcc import LCCVQE
from lcc.lp import construct_qiskit_quadratic_program

data_dir = Path('data/140')
```

# Construct quadratic program

```python
with open(data_dir/'data_0.json') as f:
    params = json.load(f)

qp = construct_qiskit_quadratic_program(params)
print(qp.prettyprint())
```

## Convert to QUBO (optional)

```python
conv = QuadraticProgramToQubo()
qubo = conv.convert(qp)
print(qubo.prettyprint())
```

# Solve using LCC

```python
lcc_solver = LCCVQE(qp, estimator=None, optimizer=None)

result = lcc_solver.solve(initial_point=None)
print(result)
```
