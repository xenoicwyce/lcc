```python
from qiskit import ClassicalRegister
from qiskit.circuit.library import TwoLocal
from lcc.lcc_functions import make_lcc_circuit
```

```python
num_qubits = 16
rotation_blocks = ["ry"]
entanglement_blocks = ["cz"]
entanglement = "circular"
reps = 1
qc = TwoLocal(num_qubits, rotation_blocks, entanglement_blocks, entanglement, reps).decompose()
qc.barrier()

measurement_operation = [0, 5, 8]
cregs = ClassicalRegister(len(measurement_operation))
qc.add_register(cregs)
qc.measure(measurement_operation, [i for i in range(len(measurement_operation))])
```

```python
qc.draw("mpl")
```

```python
lcc_qc, lcc_obs = make_lcc_circuit(qc)
print(lcc_obs)
```

```python
lcc_qc.draw("mpl")
```
